```python
import pathlib
import fitz  # PyMuPDF

folder = pathlib.Path(r"D:/Taran/OneDrive - Radboud Universiteit/Documents/jansen_rit_2003_David_own_implementation/figures/diagrams/")
out = folder / "svg"
out.mkdir(exist_ok=True)

# Parts per PDF (file name without .pdf): (left, top, right, bottom) as fractions of the page

# 0 → start at the left edge
# 0 → start at the top
# 1 → go all the way to the right edge (full width)
# 0.60 → stop 60% of the way down


splits = {
    "thalamocortical_diagram_network": {
        "diagram": (0, 0, 1, 0.65),
        "legend":  (0, 0.65, 1, 1),
    },
    "thalamocortical_diagram_nodes": {
        "thalamocortical_nodes": (0, 0, 1, 0.55),
        "legend": (0, 0.55, 1, 1),
    },
}

print("Folder exists:", folder.exists())
print("PDFs found:", [pdf.stem for pdf in folder.glob("*.pdf")])

for pdf in folder.glob("*.pdf"):
    parts = splits.get(pdf.stem)
    if parts is None:
        continue  # skip PDFs without a split definition
    for name, (fx0, fy0, fx1, fy1) in parts.items():
        page = fitz.open(pdf)[0]
        r = page.rect
        page.set_cropbox(fitz.Rect(
            r.x0 + fx0 * r.width,  r.y0 + fy0 * r.height,
            r.x0 + fx1 * r.width,  r.y0 + fy1 * r.height,
        ))
        svg = page.get_svg_image(text_as_path=True)
        (out / f"{pdf.stem}_{name}.svg").write_text(svg, encoding="utf-8")
        print(f"Saved {pdf.stem}_{name}.svg")
```
